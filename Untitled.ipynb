```python
import numpy as np
import exputils as eu
from exputils.gui.dash.experiment_data_loader import ExperimentDataLoaderWidget

# dummy functions for loading data and descriptions
def dummy_load_descriptions(experiments_directory):
    
    exp_descr = eu.AttrDict()
    for exp_idx in range(10):
        
        exp_descr['{}'.format(exp_idx)] = eu.AttrDict(
            {'id': '{}'.format(exp_idx),
             'order': exp_idx,
             'repetition_ids': [0,1,2,3,4],
             'name': 'Experiment {}'.format(exp_idx),
             'short_name': 'e{}'.format(exp_idx),
             'is_load_data': True,
             'directory': './path/to/exp{}'.format(exp_idx),
             'description': 'descr for experiment {}'.format(exp_idx)})

    return exp_descr


def dummy_load_data(exp_descriptions):
    data = dict()
    for exp_id, exp_descr in exp_descriptions.items():
        if exp_descr.is_load_data:
            data['{}'.format(exp_descr.id)] = [None]
    return data


def on_experiment_descriptions_updated_handler(event_descr):
    print("Experiment descriptions updated:")
    print(event_descr['new'])

def on_experiment_data_loaded_handler(event_descr):
    print("Experiment data loaded:")
    print(event_descr['new'])


# Create configuration for the Dash app with custom functions
config = ExperimentDataLoaderWidget.default_config()
config.load_experiment_descriptions_function.func = dummy_load_descriptions
config.load_experiment_data_function.func = dummy_load_data
config.info_area.is_visible = True

# Create the Dash app instance
exp_data_loader_app = ExperimentDataLoaderWidget(config=config)

# register event handlers
exp_data_loader_app.on_experiment_data_loaded(on_experiment_data_loaded_handler)
exp_data_loader_app.on_experiment_descriptions_updated(on_experiment_descriptions_updated_handler)

display(exp_data_loader_app)
```
