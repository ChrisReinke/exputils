# ExperimentDataLoaderWidget Example

This notebook demonstrates how to use the new `ExperimentDataLoaderWidget` with component-based architecture.

```python
# Import necessary libraries
import sys
import os
sys.path.insert(0, '/home/fuchs/data/study/projects/exputils')

import exputils as eu
import pandas as pd
from IPython.display import display
```

```python
# Create dummy data functions
def dummy_load_descriptions(experiments_directory):
    """Dummy function to load experiment descriptions"""
    descriptions = {}
    for i in range(5):
        exp_id = f"exp_{i:03d}"
        descriptions[exp_id] = {
            'id': exp_id,
            'order': i,
            'is_load_data': True,
            'name': f'Experiment {i}',
            'short_name': f'Exp{i}',
            'description': f'Description for experiment {i}',
            'directory': f'/path/to/exp_{i:03d}',
            'repetition_ids': [0, 1, 2]
        }
    return descriptions

def dummy_load_data(experiment_descriptions):
    """Dummy function to load experiment data"""
    data = {}
    for exp_id in experiment_descriptions:
        if experiment_descriptions[exp_id]['is_load_data']:
            data[exp_id] = {
                'results': pd.DataFrame({
                    'metric': [1.0, 2.0, 3.0],
                    'value': [0.1, 0.2, 0.3]
                })
            }
    return data
```

```python
# Event handlers
def on_experiment_data_loaded_handler(event):
    print(f"✅ Data loaded! {len(event.new)} experiments loaded")
    for exp_id, data in event.new.items():
        print(f"  - {exp_id}: {list(data.keys())}")

def on_experiment_descriptions_updated_handler(event):
    print(f"📝 Descriptions updated! {len(event.new)} experiments")
    for exp_id in list(event.new.keys())[:3]:  # Show first 3
        print(f"  - {exp_id}: {event.new[exp_id]['name']}")
```

```python
# Create the widget as shown in the user's example
exp_data_loader = eu.gui.dash.ExperimentDataLoaderWidget(
    load_experiment_descriptions_function=dummy_load_descriptions,
    load_experiment_data_function=dummy_load_data
)

# Register events
exp_data_loader.on_experiment_data_loaded(on_experiment_data_loaded_handler)
exp_data_loader.on_experiment_descriptions_updated(on_experiment_descriptions_updated_handler)

print(f"Widget created with {len(exp_data_loader.experiment_descriptions)} experiment descriptions")
```
